## Cuartiles, rango intercuartílico y límites para detectar valores atípicos

| Línea de código | ¿Qué hace? |
|---|---|
| `# Calculamos la medidas de posicion` | Es un comentario que introduce el cálculo de los cuartiles. |
| `Q1 = np.percentile(X, 25)` | Calcula el primer cuartil, que corresponde al percentil 25. En estos datos, `Q1 = -10.25`. |
| `Q3 = np.percentile(X, 75)` | Calcula el tercer cuartil, que corresponde al percentil 75. En estos datos, `Q3 = 10.0`. |
| `RIC = Q3 - Q1` | Calcula el rango intercuartílico, es decir, la amplitud del intervalo que abarca el 50 % central de los datos. Aquí, `RIC = 20.25`. |
| `lim_inf = Q1 - 1.5 * RIC` | Resta 1,5 veces el RIC a `Q1` para calcular el límite inferior. |
| `lim_sup = Q3 + 1.5 * RIC` | Suma 1,5 veces el RIC a `Q3` para calcular el límite superior. |
| `print("limite inferior:", lim_inf)` | Muestra el límite inferior. |
| `print("limite superior:", lim_sup)` | Muestra el límite superior. |
| `print("RIC:", RIC)` | Muestra el rango intercuartílico. |

```

### Interpretación

Según la regla de **1,5 × RIC**, los valores menores que **−40,625** o mayores que **40,375** se consideran posibles valores atípicos. En el arreglo `X`, **−100** y **−50** quedan por debajo del límite inferior, y **60** queda por encima del límite superior. Este bloque calcula e imprime los límites. Los valores atípicos se identifican al comparar cada dato con esos límites.

## Diccionario de vocabulario — Cuartiles y valores atípicos

| Término | Significado | Ejemplo de la práctica |
|---|---|---|
| Medida de posición | Valor que indica una ubicación dentro del conjunto de datos ordenado. | `Q1` y `Q3` son medidas de posición. |
| Percentil | Valor que marca una posición porcentual en los datos ordenados. Puede calcularse interpolando entre dos datos. | `np.percentile(X, 25)` calcula el percentil 25 del arreglo `X`. |
| Cuartiles | Tres valores que dividen los datos ordenados en cuatro partes. | `Q1`, `Q2` y `Q3` corresponden a los percentiles 25, 50 y 75. |
| Primer cuartil (`Q1`) | Percentil 25. Marca el extremo inferior del intervalo central de los datos. | `Q1 = −10,25`. |
| Segundo cuartil (`Q2`) | Percentil 50, también llamado mediana. | Se puede calcular con `np.percentile(X, 50)`. En `X`, vale `0,5`. |
| Tercer cuartil (`Q3`) | Percentil 75. Marca el extremo superior del intervalo central de los datos. | `Q3 = 10,0`. |
| Rango intercuartílico (`RIC`) | Diferencia entre el tercer y el primer cuartil. Mide la amplitud del intervalo que contiene aproximadamente el 50 % central de los datos. | `RIC = Q3 − Q1 = 20,25`. |
| Regla de `1,5 × RIC` | Criterio para identificar posibles valores atípicos mediante dos límites calculados a partir de los cuartiles. | `1,5 × 20,25 = 30,375`. |
| Límite inferior (`lim_inf`) | Umbral que se obtiene al restar `1,5 × RIC` a `Q1`. | `−40,625`. |
| Límite superior (`lim_sup`) | Umbral que se obtiene al sumar `1,5 × RIC` a `Q3`. | `40,375`. |
| Valor atípico | Dato que, según esta regla, queda por debajo del límite inferior o por encima del superior. | `−100`, `−50` y `60`. |
