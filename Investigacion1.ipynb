## Objetivos

Explicar qué es el descenso de gradiente, cómo funciona y por qué es el mecanismo con el que aprende la mayoría de los modelos de inteligencia artificial actuales, apoyándose en fuentes verificables y en dos ejemplos: uno conceptual (el precio de una casa) y uno numérico (el excursionista ciego, calculado en Excel).


## Antecedentes/Escenario

Entrenar un modelo de aprendizaje automático consiste en ajustar sus parámetros para que una función de pérdida (el error entre lo que el modelo predice y lo esperado) sea lo más pequeña posible. El descenso de gradiente es uno de los algoritmos de optimización más populares y, según Ruder (2016), la forma más común de optimizar redes neuronales.

Su origen suele atribuirse a Augustin-Louis Cauchy, quien lo presentó el 18 de octubre de 1847 en la Academia de Ciencias de Francia para resolver sistemas de ecuaciones en cálculos astronómicos, como la órbita de un cuerpo celeste (Cauchy, 1847; Lemaréchal, 2012). Robbins y Monro (1951) sentaron las bases de los métodos estocásticos de aproximación, antecedente del descenso de gradiente estocástico. En 1986, Rumelhart et al. mostraron que la retropropagación de errores permite ajustar los pesos de redes con capas ocultas minimizando el error. Hoy, optimizadores derivados como Adam (Kingma y Ba, 2015) entrenan modelos como el Transformer (Vaswani et al., 2017) y GPT-3 (Brown et al., 2020).

## Planteamiento del problema

Un modelo de IA puede tener millones o miles de millones de parámetros (GPT-3, por ejemplo, tiene 175 mil millones; Brown et al., 2020). Ajustarlos a mano es imposible. La pregunta es: ¿cómo encontrar, de forma automática y eficiente, los valores de los parámetros que minimizan el error? Además: ¿qué decisiones del método, como el tamaño del paso o la variante usada, determinan que el modelo aprenda bien, lento o no aprenda?


## Pasos por realizar

### Descenso de gradiente y su influencia en los modelos de IA

El descenso de gradiente es un método iterativo que mide qué tan mal predice el modelo mediante una función de costo, de pérdida o de error: una fórmula matemática que cuantifica qué tan equivocado está el modelo al hacer una predicción (Jordan, 2017). Esta función compara el valor que se desea obtener ($y$) con el valor que el modelo ha pronosticado ($\hat{y}$; Goodfellow et al., 2016).

Si la predicción es pésima, la función arroja un número alto (un castigo o costo); en cambio, si la predicción es muy exacta, el valor se acerca a cero. Los ejemplos más comunes son:

- **Error cuadrático medio (MSE):** se utiliza en regresión; eleva al cuadrado las diferencias entre lo real y lo predicho para penalizar más los errores grandes (Goodfellow et al., 2016).
- **Entropía cruzada:** muy utilizada en clasificación, por ejemplo, para distinguir si una imagen es un gato o un perro (Goodfellow et al., 2016).

El procedimiento general del descenso de gradiente (Goodfellow et al., 2016; Ruder, 2016) es el siguiente:

1. **Paso 1:** inicializar los parámetros $\theta$ con valores arbitrarios, normalmente aleatorios.
2. **Paso 2:** calcular la pérdida $L(\theta)$ con los datos de entrenamiento.
3. **Paso 3:** calcular el gradiente $\nabla L(\theta)$, que apunta en la dirección en la que el error crece más rápido.
4. **Paso 4:** actualizar los parámetros en sentido contrario al gradiente: $\theta \leftarrow \theta - \alpha \cdot \nabla L(\theta)$, donde $\alpha$ es la tasa de aprendizaje (tamaño del paso).
5. **Paso 5:** repetir los pasos 2 a 4 hasta que la pérdida deje de disminuir de forma apreciable.


## Desarrollo

Para explicar la idea de manera lúdica existe un ejemplo clásico: una persona está en la cima de una montaña y su objetivo es descender hasta la base lo más pronto posible; como los temas en inteligencia artificial no son fáciles, la persona es ciega y se pronostica una tormenta. ¿Qué debería hacer? La lógica indica que debe sentir con los pies la pendiente del terreno y avanzar hacia donde este desciende. Se le presentan dos alternativas: dar pasos muy cortos, con lo que se demoraría una eternidad, o dar pasos muy largos, con lo que se pasaría de la base. Esta intuición coincide con la que explica Sanderson (2017): moverse cuesta abajo, donde un paso grande avanza más rápido, pero puede pasarse del mínimo.

En términos matemáticos, el proceso consiste en:

- calcular el gradiente (pendiente o derivada) de la función de costo respecto a los parámetros del modelo;
- actualizar los parámetros moviéndose en la dirección opuesta al gradiente (pendiente negativa); y
- controlar el tamaño de cada paso mediante un valor llamado tasa de aprendizaje (*learning rate*): si es muy pequeño, el aprendizaje se demora demasiado; si es muy grande, se pasa del objetivo (Sanderson, 2017).

En resumen, el descenso de gradiente (*gradient descent*) es el motor de optimización que permite a los modelos de inteligencia artificial aprender de sus errores, mientras que la función de costo (*cost function*) es la métrica que cuantifica qué tan grandes son esos errores. Como regla general, si el costo medido sobre los datos de entrenamiento es alto, el modelo es malo y sus predicciones fallan; si es cercano a cero, el modelo se ajusta bien a esos datos.

Otro ejemplo análogo, antes de entrar en las matemáticas: imagina que se quiere estimar el precio de una casa a partir de su tamaño. Si la casa cuesta 100 000 dólares y el modelo predice 90 000, el error es de 10 000 dólares; si predice 150 000, el error es de 50 000. La función de costo toma los errores de miles de ejemplos, los eleva al cuadrado o los promedia y entrega una penalización total. El objetivo del modelo es hacer que esa penalización sea lo más pequeña posible.

El descenso de gradiente y las funciones de costo son la base del entrenamiento de las redes neuronales y de los modelos de lenguaje actuales (Ruder, 2016), y su influencia se resume en tres puntos clave:

- **Automatización del aprendizaje:** el programador no necesita escribir miles de reglas lógicas a mano; el descenso de gradiente permite que el modelo descubra las reglas por sí mismo, ajustando millones de variables de forma automática (Goodfellow et al., 2016).
- **Eficiencia con datos masivos:** variantes como el descenso de gradiente estocástico (SGD) permiten entrenar modelos con enormes volúmenes de imágenes o textos, procesando los datos en pequeños grupos (lotes o *batches*) sin saturar la memoria de las computadoras (Bottou et al., 2018; Ruder, 2016).
- **Especialización del modelo:** según la función de costo que se elija, el descenso de gradiente moldea la IA para tareas muy distintas (Goodfellow et al., 2016):
  - con el error cuadrático medio, la IA aprende a predecir números (regresión, como el clima o los precios);
  - con la entropía cruzada, la IA aprende a clasificar (por ejemplo, identificar si una foto es un gato o un perro).


### Fundamento

Goodfellow et al. (2016, cap. 4.3) expresan la actualización como $x' = x - \epsilon \nabla_x f(x)$, con $\epsilon$ como tasa de aprendizaje. Ya en 1847, Cauchy mostró que, para un $\epsilon > 0$ pequeño, $f(x - \epsilon \nabla f) \approx f(x) - \epsilon \lVert \nabla f \rVert^2$: dar un paso contra el gradiente reduce el valor de la función (Lemaréchal, 2012). En la notación de este trabajo, la regla de actualización se escribe como en la ecuación 1, donde $\theta$ son los parámetros del modelo, $\alpha$ la tasa de aprendizaje y $\nabla L(\theta)$ el gradiente de la pérdida.

$$\theta \leftarrow \theta - \alpha \, \nabla L(\theta) \tag{1}$$


### Variantes

Ruder (2016) distingue tres: **por lotes** (usa todo el conjunto de datos en cada paso: preciso pero costoso con muchos datos), **estocástico o SGD** (un ejemplo por actualización: rápido pero ruidoso) y **por minilotes** (un subconjunto pequeño, con tamaños habituales de 50 a 256), que es la opción habitual para entrenar redes neuronales. **Adam** (Kingma y Ba, 2015) añade tasas de aprendizaje adaptativas por parámetro, calculadas a partir de estimaciones de los momentos de primer y segundo orden del gradiente, con bajo consumo de memoria.

### Influencia en los modelos de IA

- **Redes neuronales:** la retropropagación (Rumelhart et al., 1986) calcula cómo cambia el error respecto de cada peso y el descenso de gradiente lo usa para actualizarlos. Así, las capas ocultas aprenden representaciones internas de la tarea.
- **Transformer:** se entrenó con Adam ($\beta_1 = 0{,}9$; $\beta_2 = 0{,}98$; $\epsilon = 10^{-9}$) y una tasa de aprendizaje que aumenta durante 4000 pasos y luego decrece (Vaswani et al., 2017).
- **GPT-3:** se entrenó con Adam ($\beta_1 = 0{,}9$; $\beta_2 = 0{,}95$; $\epsilon = 10^{-8}$), recorte del gradiente global en 1,0 y decaimiento coseno de la tasa de aprendizaje (Brown et al., 2020).

En todos los casos, la tasa de aprendizaje y su calendario son decisiones de ingeniería que condicionan si el modelo converge, como muestra el ejemplo práctico (figuras 1 y 2).

### Limitaciones

En las redes neuronales profundas no se puede garantizar que el entrenamiento encuentre la mejor solución posible, es decir, el mínimo global. Según Dauphin et al. (2014), cuando se trabaja con muchas variables, uno de los principales obstáculos son los **puntos de silla**: zonas donde la función parece casi plana, aunque puede subir en unas direcciones y bajar en otras, lo que puede hacer que el aprendizaje avance muy lentamente.

- **Caso 1 ($\alpha = 0{,}1$):** $x$ se multiplica por 0,8 en cada paso ($1 - 2\alpha$). El error baja con seguridad, pero despacio: de 16,00 a 0,18 en 10 pasos.
- **Caso 2 ($\alpha = 0{,}5$):** $x$ se multiplica por 0 y llega a 0 en un solo paso; esto es particular de esta función y no una regla general.
- **Caso 3 ($\alpha = 1{,}1$):** $x$ se multiplica por $-1{,}2$. Como el factor tiene valor absoluto mayor que 1, el error crece (de 16 a 613,4 en 10 pasos) y el entrenamiento diverge.

El ejemplo confirma lo señalado en el desarrollo: la tasa de aprendizaje determina si el modelo aprende bien, lento o no aprende.


## Conclusiones

- El descenso de gradiente, propuesto por Cauchy en 1847, es el mecanismo básico con el que se ajustan los parámetros de los modelos de IA: se mueve cada parámetro en el sentido que reduce el error (Lemaréchal, 2012).
- Su influencia es directa: la retropropagación calcula el gradiente y variantes como el descenso estocástico por minilotes y Adam permiten entrenar modelos como el Transformer y GPT-3.
- La calidad del entrenamiento depende de la tasa de aprendizaje, como se comprobó en el ejemplo práctico (convergencia lenta con $\alpha = 0{,}1$ y divergencia con $\alpha = 1{,}1$), y de la forma de la función de error, donde los puntos de silla son un obstáculo importante en alta dimensión (Dauphin et al., 2014).
