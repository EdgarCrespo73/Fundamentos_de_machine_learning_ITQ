```python
# Importamos librerias para la creacion de datos 
import numpy as np
import pandas as pd
import random
random.seed (42)
import matplotlib.pyplot as plt
from scipy import stats
from sklearn import datasets
from scipy.stats import skew, kurtosis
from pprint import pprint
```

```python
#Creo la estructura de datos
nivel_estudios = []
nivel_estudios.extend(["Secundaria"]*110)
nivel_estudios.extend(["Bachillerato"]*172)
nivel_estudios.extend(["Grado"]*145)
nivel_estudios.extend(["Master"]*52)
nivel_estudios.extend(["Doctorado"]*21)
nivel_estudios = np.random.permutation(nivel_estudios)
```

```python
#Imprimir el conjunto de datos
print(nivel_estudios)
```

```python
estudios, frecuencia = np.unique(nivel_estudios,return_counts =True)
estudios, frecuencia = np.unique(nivel_estudios, return_counts=True)

print("Estudios:", estudios)
print("Frecuencia:", frecuencia)
```

```python
#realizamos un diagrama de barras para entender como estan los datos

plt.bar(estudios, frecuencia)

# Etiquetar los ejes y colocar el título
plt.xlabel("Nivel de estudios")
plt.ylabel("Frecuencia absoluta")
plt.title("Niveles de estudios en la ciudad de Valencia")

# Mostrar el gráfico
plt.show()
```

```python
#Dibujamos un diagrama de sectores o pie
# Dibujamos un diagrama de sectores
colores = ["yellow", "red", "green", "blue", "magenta"]

plt.pie(
    frecuencia,
    labels=estudios,
    colors=colores,
    autopct="%.2f%%"
)
plt.title("Distribución de los niveles de estudios en Valencia")
plt.show()
```

```python
df_estudios = pd.DataFrame({
    "Nivel de estudios": estudios,
    "Frecuencia absoluta": frecuencia
})

df_estudios["Porcentaje (%)"] = (
    df_estudios["Frecuencia absoluta"] / df_estudios["Frecuencia absoluta"].sum() * 100
).round(2)

display(df_estudios)
    
```

```python
plt.pie(
    df_estudios["Frecuencia absoluta"],
    labels=df_estudios["Nivel de estudios"],
    autopct="%.2f%%"
)
plt.title("Distribución de los niveles de estudios en Valencia")
plt.show()
```
