```python
from sklearn.model_selection import KFold
# Importa la clase KFold de scikit-learn, que divide los datos en K particiones (pliegues) para la validación cruzada.
```

```python
X = ["a", "b", "c", "d", "e", "f", "g", "h", "i", "j"]
# Crea una lista con 10 letras que representan las muestras (datos) que se van a dividir.
```

```python
# Crear 5 particiones con asignación aleatoria.
# (Comentario original: describe el objetivo de las siguientes líneas.)
kf = KFold(n_splits=5, shuffle=True)
# Crea el objeto KFold con 5 pliegues; shuffle=True mezcla los datos al azar antes de dividirlos.
bolsas = kf.split(X)
# Genera los índices de entrenamiento y de prueba para cada pliegue (devuelve un generador, aún no se ejecuta).
print(bolsas)
# Muestra en pantalla el generador (solo el objeto, no los índices).
```

```python
k = 1
# Inicializa el contador k que numera las iteraciones.

for train, test in bolsas:
    # Recorre cada pliegue; en cada vuelta train recibe los índices de entrenamiento y test los de prueba.
    print("Iteración", k, ":")
    # Imprime el número de la iteración actual.
    print(" - Entrenamiento:", train)
    # Imprime los índices de las muestras usadas para entrenar.
    print(" - Test:", test)
    # Imprime los índices de las muestras usadas para probar (validar).
    k = k + 1
    # Aumenta el contador en 1 para la siguiente iteración.
```

## Diccionario

| Término | Significado |
|---|---|
| **Validación cruzada (cross-validation)** | Técnica que divide los datos en varias partes y rota cuál se usa para probar, para evaluar un modelo de forma más confiable. |
| **K-Fold** | Método de validación cruzada que crea K particiones (pliegues) del conjunto de datos. |
| **Pliegue (fold) / bolsa** | Cada una de las K particiones en que se divide el conjunto de datos. |
| **sklearn (scikit-learn)** | Biblioteca de Python para aprendizaje automático. |
| **model_selection** | Módulo de scikit-learn con herramientas para dividir datos y evaluar modelos. |
| **KFold** | Clase de scikit-learn que genera los índices de los K pliegues. |
| **import / from ... import** | Instrucción que trae una librería o clase para usarla en el programa. |
| **n_splits** | Parámetro que indica en cuántos pliegues se divide el conjunto de datos. |
| **shuffle** | Parámetro que, si es True, mezcla los datos al azar antes de dividirlos. |
| **split()** | Método que genera, para cada pliegue, los índices de entrenamiento y de prueba. |
| **Generador** | Objeto que entrega sus valores uno a uno al recorrerlo, sin calcularlos todos de antemano. |
| **X** | Variable que contiene las muestras (datos) del ejemplo. |
| **Lista** | Estructura de Python que guarda varios elementos ordenados, entre corchetes. |
| **Índice** | Posición numérica de un elemento dentro de la lista (empieza en 0). |
| **Entrenamiento (train)** | Conjunto de muestras con las que el modelo aprende. |
| **Prueba (test)** | Conjunto de muestras reservadas para evaluar el modelo. |
| **Iteración** | Cada repetición del ciclo; aquí, cada pliegue evaluado. |
| **for** | Ciclo que repite un bloque de código por cada elemento de una colección. |
| **print()** | Función que muestra texto o valores en pantalla. |
| **Contador (k)** | Variable numérica que aumenta en cada vuelta para llevar la cuenta. |
| **Comentario (#)** | Línea que Python ignora; sirve para explicar el código. |
