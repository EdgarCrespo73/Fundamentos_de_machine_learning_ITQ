```python
#Carga de datos
datos = datasets.load_iris()
print(np.shape(datos.data))
```

| Línea | Qué hace |
|---|---|
| `#Carga de datos` | Es un comentario que identifica el propósito de la celda. Python lo ignora al ejecutar el código. |
| `datos = datasets.load_iris()` | Carga Iris y lo guarda en `datos`. `datos.data` contiene las medidas de las flores y `datos.target` contiene la especie correspondiente a cada fila, codificada como `0`, `1` o `2`. |
| `print(np.shape(datos.data))` | Imprime el número de filas y columnas de las medidas. También puedes escribir `print(datos.data.shape)`. |

```python
# Extrar la matriz de datos X y la variable target Y
X = datos.data
y = datos.target
```

| Línea | Qué hace | Dimensiones |
|---|---|---|
| `X = datos.data` | Guarda las cuatro medidas de cada flor. Cada fila representa una flor y cada columna representa una medida del sépalo o del pétalo. | `(150, 4)` |
| `Y = datos.target` | Guarda la especie de cada flor como un número. En Iris, `0` es setosa, `1` es versicolor y `2` es virginica. | `(150,)` |

```python
# Particion externa test: hold-out split 80-20%.
X_train, X_test, y_train , y_test = train_test_split(datos.data,datos.target,test_size=0.2,random_state=42)
print("Train dimensions:",np.shape(X_train))
print("Test dimensions:" ,np.shape(X_test))
```

| Línea | Qué hace |
|---|---|
| `# Particion externa test: hold-out split 80-20%.` | Es un comentario que describe la división entre entrenamiento y prueba. |
| `X_train, X_test, y_train, y_test = train_test_split(...)` | Divide las medidas (`datos.data`) y las especies (`datos.target`) juntas, de modo que cada flor conserva su etiqueta. `test_size=0.2` reserva el 20 % para prueba y `random_state=42` permite reproducir la misma división. |
| `print("Train dimensions:", np.shape(X_train))` | Muestra las dimensiones de las medidas de entrenamiento. |
| `print("Test dimensions:", np.shape(X_test))` | Muestra las dimensiones de las medidas de prueba. |

```python
#Estandartizar los datos de entrenamiento y test
standartizer = StandardScaler()
```

standartizer = StandardScaler()	Crea el objeto que estandarizará las cuatro medidas de Iris. scaler es el nombre que le damos a la variable.

```python
alg = Pipeline([
    ("scaler", StandardScaler()),
    ("svc", SVC(C=0.1, gamma="auto", kernel="rbf")),
])
```

| Elemento | Qué hace |
|---|---|
| `("scaler", StandardScaler())` | Primer paso. Estandariza las cuatro medidas de las flores. |
| `("svc", SVC(...))` | Segundo paso. Usa las medidas estandarizadas para aprender a clasificar las flores por especie. |
| `C=0.1` | Aumenta la regularización respecto a `C=1`. El modelo puede tolerar más errores de entrenamiento a cambio de un margen más amplio. |
| `gamma="auto"` | Fija `gamma` en `1 / número de características`. Como Iris tiene cuatro medidas, aquí vale `0.25`. |
| `kernel="rbf"` | Permite que el modelo aprenda fronteras curvas entre las especies. |

```python
# Definimos el algoritmo 
alg = SVC (C=0.1,gamma="auto", kernel ="rbf")
```

| Parte | Qué hace |
|---|---|
| `alg =` | Guarda el clasificador en la variable `alg`. |
| `SVC(...)` | Crea un clasificador basado en máquinas de vectores de soporte. |
| `C=0.1` | Da menos peso a los errores de entrenamiento que `C=1`, lo que favorece un margen de separación más amplio. |
| `gamma="auto"` | Calcula `gamma` como `1 / número de características`. En Iris vale `1 / 4 = 0.25`. Con el kernel RBF, `gamma` controla qué tan rápido disminuye la influencia de una flor a medida que otra se aleja en el espacio de las medidas. |
| `kernel="rbf"` | Permite aprender fronteras curvas para separar las especies. |

```python
#Cross validation interno
cv = StratifiedKFold(n_splits=5,shuffle=True,random_state=42)
#Utilizar comando croos_val_predict para extraer directamente las predicciones de las 5 bolsas de validacion.
y_pred=cross_val_predict(alg,X_train,y_train,cv=cv)
pprint(y_pred)
pprint(y_train)
```

```python
# Evaluar las predicciones de la validación cruzada
exactitud = metrics.accuracy_score(y_train, y_pred)
precision = metrics.precision_score(y_train, y_pred, average="macro")
sensibilidad = metrics.recall_score(y_train, y_pred, average="macro")
f1 = metrics.f1_score(y_train, y_pred, average="macro")

print("Resultados de la validación cruzada")
print(f"Exactitud     {exactitud:.2%}")
print(f"Precisión     {precision:.2%}")
print(f"Sensibilidad  {sensibilidad:.2%}")
print(f"F1-score      {f1:.2%}")
```

| Métrica | Qué pregunta responde |
|---|---|
| **Exactitud** (`accuracy`) | De las 120 flores, ¿qué porcentaje clasificó correctamente el modelo? |
| **Precisión** (`precision`) | De las flores que el modelo predijo como una especie, ¿qué porcentaje realmente pertenecía a esa especie? |
| **Sensibilidad** (`recall`) | De todas las flores que realmente pertenecían a una especie, ¿qué porcentaje logró identificar el modelo? |
| **F1-score** | ¿Qué tan bien combina el modelo la precisión y la sensibilidad para cada especie? El F1 baja si alguna de las dos es baja. |

```python
#Extraemos la matriz de confusion 

print("Matriz de confusión\n", metrics.confusion_matrix(y_train, y_pred))

```

**Cómo leer la matriz de confusión**
- Filas → especie real.
- Columnas → especie predicha.
- Diagonal principal → aciertos. Por ejemplo, la celda [1, 1] cuenta las flores versicolor clasificadas correctamente.
- Fuera de la diagonal → errores. Por ejemplo, la celda [1, 2] cuenta las flores versicolor que el modelo predijo como virginica.
- En Iris, el orden es 0 = setosa, 1 = versicolor, 2 = virginica.
  



```python
print("Tabla de métricas:\n", metrics.classification_report(y_train, y_pred))
```

**Cómo leer la tabla de métricas** El reporte muestra precision, recall y f1-score para cada especie. La columna support indica cuántas flores reales de esa especie hay en y_train. La fila macro avg da el mismo peso a cada especie y weighted avg pondera cada especie según su cantidad de flores.

Agregué target_names=datos.target_names para que aparezcan los nombres de las especies y digits=3 para mostrar tres decimales.

```python
# Crear y entrenar el modelo definitivo
model = clone(alg) #Definimos el modelo con los hiperparametros optimos
model.fit(X_train, y_train)# Entrenamos el modelos con los datos train y val

# finalmente evaluamos el test con el metodo score()
test_results = model.score(X_test, y_test)

print(f"Exactitud en test: {test_results:.4%}")
```

```python
# Predecir las especies de las flores de prueba
y_pred_test = model.predict(X_test)

# Mostrar las métricas por especie
print("Métricas de la prueba final")
print(metrics.classification_report(
    y_test,
    y_pred_test,
    labels=[0, 1, 2],
    target_names=datos.target_names,
    digits=3,
))

# Mostrar los aciertos y errores por especie
print("Matriz de confusión")
print(metrics.confusion_matrix(y_test, y_pred_test, labels=[0, 1, 2]))
```

- **Reporte de clasificación.** Muestra la precisión, la sensibilidad y el F1 por especie. support indica cuántas flores de cada especie había en la prueba.
- Matriz de confusión. Las filas representan las especies reales y las columnas, las predichas. En ambos ejes, el orden es setosa, versicolor y virginica.
Con el 100 % de exactitud que obtuviste, los 30 aciertos deben aparecer en la diagonal de la matriz y todas las demás celdas deben tener un cero.
