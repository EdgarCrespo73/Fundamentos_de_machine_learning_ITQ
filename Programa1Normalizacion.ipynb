**Normalización:** Proceso en el cual se convierten los datos al rango [0, 1]; este proceso se denomina HOMOTECIA.

**Fórmula:** $$
X_{norm} = \frac{X - X_{min}}{X_{max} - X_{min}}
$$



```python
from sklearn import preprocessing
import numpy as np
```

**preprocessing:** Transforma y limpia los datos brutos antes de introducirlos en un modelo de Machine Learning.

**numpy:** Biblioteca de código abierto para Python que sirve para realizar cálculos matemáticos, 
científicos y de análisis de datos de manera sumamente rápida y eficiente.

```python
# Carga de datos.
X_train = np.array ([
    [ 1., -4., 5.],
    [ 5., 0., 2.],
    [ 0., 6., -3.]
])
print(X_train)
```

```python
# Limpieza de datos: normalización.
normalizer = preprocessing.MinMaxScaler()
X_train_norm = normalizer.fit_transform(X_train)
```

```python
# Salida de resultados.
print(X_train_norm)
```

**MinMaxScaler:** Técnica de preprocesamiento de datos en machine learning
que transforma los valores de una característica numérica para que 
queden acotados dentro de un rango específico y predefinido, habitualmente entre 0 y 1.
Aplica una transformación lineal a cada valor (X) restándole el mínimo de la variable y 
dividiéndolo por el rango total (máximo menos mínimo).

**.fit_transform:** Función que calcula los parámetros estadísticos de tus datos y los transforma
inmediatamente en un solo paso eficiente.
En lugar de llamar a dos funciones por separado,
este método combina los dos procesos esenciales del preprocesamiento: ajustar (fit) y transformar (transform).


```python
# Ejemplo para ver que los resultados son iguales si aplicamos los métodos
# "fit" y "transform" por separado
X_train_2 = np.copy(X_train)
# Crea un objeto "fitted_normalizer" llamando al método "fit"
f = normalizer.fit(X_train)
# Aplica el objeto sobre los datos "X_train_2" para obtener "X_train_norm_2"
X_train_norm_2 = f.transform(X_train)
# Comprueba si "X_train_norm" y "X_train_norm_2" son iguales
print(np.array_equiv(X_train_norm, X_train_norm_2))
```

```python
# Normalizar un nuevo conjunto de datos de test (SOLO HAY QUE APLICAR EL MÉTODO
# "transform", NO EL "fit")
X_test = np.array([[ 1., -3., 2.],
[ 3., 0., 0.],
[ 0., 4., -1.]])
# Normalizar los datos de "X_test"
X_test_norm=f.transform(X_test)
```
