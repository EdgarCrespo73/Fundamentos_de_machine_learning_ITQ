```python
import numpy as np
from sklearn import datasets
from sklearn.dummy import DummyClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler


```

**from sklearn import datasets**
Importa el módulo datasets de scikit-learn. Puedes usarlo para cargar conjuntos de datos de ejemplo. Por ejemplo, datasets.load_iris() carga datos con medidas de flores y la especie de cada una.

**from sklearn.dummy import DummyClassifier**
Importa un clasificador sencillo que sirve como referencia para evaluar otros modelos. Por ejemplo, con strategy="most_frequent", siempre predice la clase más común del conjunto de entrenamiento. Si 70 de 100 ejemplos son de la clase A, siempre predice A.

Importa una función que divide los datos en dos grupos. Uno se usa para entrenar el modelo y el otro para evaluar cómo predice con ejemplos que no vio durante el entrenamiento. Por ejemplo, test_size=0.2 reserva aproximadamente el 20 % para la prueba.

**from sklearn.preprocessing import StandardScaler** importa la clase StandardScaler de scikit-learn para que puedas usarla en tu código. Esa línea solo hace la importación.

**StandardScaler** estandariza cada columna de medidas. Resta la media de la columna y divide por su desviación estándar. Así, en Iris, las medidas de pétalos y sépalos quedan en una escala comparable.



```python
# Carga de datos.
iris = datasets.load_iris()
print(iris.target_names)
```

**iris = datasets.load_iris()** se lee así: carga el conjunto de datos Iris y guárdalo en la variable iris.
- datasets.load_iris() llama a la función de scikit-learn que devuelve los datos.
- iris = asigna ese resultado a la variable iris para usarlo después.
**print(iris.target_names)** Imprime el nombre de las 3 variedades.


```python
# Mostrar características de la tabla de datos.
print("Tabla de datos: %d instancias y %d atributos" % (iris.data.shape[0],
iris.data.shape[1]))
print("Valores de la clase:", set(iris.target))
# Cuantificamos el numero de instancias que contine el data set por clase.
valores,ocurrencias = np.unique(iris.target,return_counts=True)
print(valores,ocurrencias)
```

# Mostrar características de la tabla de datos.
Este código muestra cuántas flores hay, cuántas medidas tiene cada flor y qué códigos representan a las especies.
**iris.data.shape[0]** cuenta las filas. Cada fila corresponde a una flor.
**iris.data.shape[1]** cuenta las columnas. Cada columna corresponde a una medida.
Los dos %d se sustituyen por esos números enteros, en el mismo orden.
**set(iris.target)** elimina los códigos repetidos y muestra los tres valores distintos
. **0 corresponde a setosa,** **1 a versicolor** y **2 a virginica**.
# Cuantificamos el numero de instancias que contine el data set por clase.
**np.unique(iris.target, return_counts=True)** devuelve dos arreglos. valores guarda los códigos de especie distintos y ocurrencias guarda cuántas veces aparece cada código, en el mismo orden.
**print(valores, ocurrencias)** muestra ambos arreglos:

```python
# Validación: hold-out split 80-20%. #Particion externa

X_training, X_test, y_training, y_test = train_test_split(
    iris.data, iris.target, test_size=0.2, random_state=42
)
valores_test,ocur_test=np.unique(y_test,return_counts= True)
print("Test:","clases:",valores_test,"ocurrencias:",ocur_test)

# Partición interna: reserva el 20 % de las 120 flores restantes para validación.
#X_train, X_val, y_train, y_val = train_test_split(
 #   X_training, y_training, test_size=0.2, random_state=42
#)

#valores_train, ocur_train = np.unique(y_train, return_counts=True)
#print("Entrenamiento:", "clases:", valores_train, "ocurrencias:", ocur_train)

#valores_val, ocur_val = np.unique(y_val, return_counts=True)
#print("Validación:", "clases:", valores_val, "ocurrencias:", ocur_val)
```

```python
# Estandarizar las caracteristicas de entreanamiento y de test
standardizer = StandardScaler()
X_training = standardizer.fit_transform(X_training)
X_test = standardizer.transform(X_test)
```

Este código estandariza las cuatro medidas de las flores de Iris. Para cada columna, resta su media y divide por su desviación estándar.
**1. standardizer = StandardScaler()** crea el objeto que hará la transformación.
**2. X_training = standardizer.fit_transform(X_training)** calcula la media y la desviación estándar de cada columna usando las 120 flores de X_training. Luego transforma esas medidas y guarda el resultado en la misma variable.
**3. X_test = standardizer.transform(X_test)** transforma las medidas de las 30 flores de prueba usando las medias y desviaciones ya calculadas. Así, los datos de prueba quedan en la misma escala y se mantienen fuera del ajuste.

```python
# Validación: hold-out split 80-20%. #Particion interna
# Partición interna: 80 % para entrenamiento y 20 % para validación.
X_train, X_val, y_train, y_val = train_test_split(
    X_training, y_training, test_size=0.2, random_state=42
)

valores_train, ocur_train = np.unique(y_train, return_counts=True)
print("Entrenamiento:", "clases:", valores_train, "ocurrencias:", ocur_train)

valores_val, ocur_val = np.unique(y_val, return_counts=True)
print("Validación:", "clases:", valores_val, "ocurrencias:", ocur_val)

```

**1. train_test_split(...)** divide las 120 flores de la partición externa en 96 para entrenamiento y 24 para validación. Las variables X guardan las medidas y las variables y guardan las especies. Usé random_state=42, como en tu partición anterior, porque el valor aparece cortado en la captura.

**2. np.unique(..., return_counts=True)** obtiene los códigos de especie y cuenta cuántas flores hay de cada una. Los dos print() muestran esos conteos para entrenamiento y validación.

**3. DummyClassifier(strategy="prior")** crea un clasificador de referencia. Cuando lo entrenes con clf.fit(X_train, y_train), identificará la especie más frecuente y la usará para todas sus predicciones.
Si vas a estandarizar las medidas, ejecuta primero esta división y ajusta StandardScaler solo con X_train.


```python
# Construcción del objeto que contiene el algoritmo de aprendizaje.
clf = DummyClassifier(strategy="prior", random_state=42)
```

**clf = DummyClassifier()** crea un clasificador de referencia y lo guarda en la variable clf. Sirve para comparar tu modelo con una regla simple.

Cuando ejecutas **clf.fit(X_train, y_train)**, el clasificador identifica la especie más frecuente en y_train. Con la configuración predeterminada, luego predice esa misma especie para todas las flores, independientemente de sus medidas.

```python
# Entrenamiento del algoritmo de aprendizaje.
clf.fit(X_train, y_train)
```

**clf.fit(X_train, y_train)** ajusta el clasificador con el conjunto de entrenamiento. X_train contiene las cuatro medidas de cada flor y y_train contiene su especie.

Como clf es un **DummyClassifier()** con la configuración predeterminada, durante el ajuste cuenta cuántas veces aparece cada especie en y_train e identifica la más frecuente. Cuando luego llames a **clf.predict(X_test)**, predecirá esa especie para todas las flores de prueba, independientemente de sus medidas.

```python
# Evaluación del algoritmo de aprendizaje con el metodo score que devuelve direcamente la metrica accuracy.
val_accuracy = clf.score(X_val, y_val)
print("Exactitud en validacion:", val_accuracy)
test_accuracy = clf.score(X_test, y_test)
print("Exactitud en test:",test_accuracy )
```

Este código calcula la exactitud del clasificador en dos conjuntos. La exactitud es la proporción de predicciones correctas.
**- val_accuracy = clf.score(X_val, y_val)** predice las especies de las 24 flores de validación y compara las predicciones con sus especies reales (y_val).

**- test_accuracy = clf.score(X_test, y_test)** hace lo mismo con las 30 flores de prueba.

**- Cada print()** muestra el resultado correspondiente. Por ejemplo, 0.20 significa 20 % de aciertos. Lo que es un valor muy bajo, validacion en test obtien un 36%, no apto para produccion.

```python
# Predicciones para el conjunto de validación
y_pred_val = clf.predict(X_val)
print("Predicciones de validación:", y_pred_val)
print("Etiquetas reales de validación:", y_val)

# Predicciones para el conjunto de prueba
y_pred_test = clf.predict(X_test)
print("\nPredicciones de prueba:", y_pred_test)
print("Etiquetas reales de prueba:", y_test)
```

**Demostrado que el modelo matematico utilizado, no es para nada preciso**.

```python
#Aplicamos un modelo de calsificador mas complejo dummyclassifier
from sklearn.svm import SVC

svc = SVC(C=0.5) #Definimos algoritmo
svc.fit(X_train, y_train)# Entranamos modelo

# Evaluar en validacion
val_accuracy = svc.score(X_val,y_val) # Evaluamos modelo de validacion 
print("Exactitud en validacion", np.round(test_accuracy*100,4),"%")

#Evaluar en prueba
test_accuracy = svc.score(X_test,y_test) #Evaluamos modelo en test
print("Exactitud en test", np.round(test_accuracy*100, 4),"%")
```

**En este modelo cambia draticamente , exactitud en validacion y test alcanza el 96.67%**

**from sklearn.svm import SVC** importa la clase SVC de scikit-learn.

**svc = SVC(C=0.5)** crea el clasificador y lo guarda en svc. El parámetro C controla cuánto penaliza los errores durante el entrenamiento.

**svc.fit(X_train, y_train)** entrena el modelo. X_train contiene las medidas de las flores y y_train sus especies reales.

**val_accuracy = svc.score(X_val, y_val)** predice las especies de las flores de validación, las compara con sus especies reales y guarda la proporción de aciertos.

**print("Exactitud en validación:", np.round(val_accuracy * 100, 4), "%")** convierte esa proporción en porcentaje, la redondea a cuatro decimales y la muestra. Por ejemplo, 0.95 equivale al 95 %.

**test_accuracy = svc.score(X_test, y_test)** calcula la proporción de aciertos en las flores del conjunto de prueba.

**print("Exactitud en prueba:", np.round(test_accuracy * 100, 4), "%")** muestra la exactitud de prueba como porcentaje.

# Diccionario técnico del ejercicio Iris

## Datos y su representación

| Término | Significado en el ejercicio |
|---|---|
| **Conjunto de datos (`dataset`)** | Colección de ejemplos. El conjunto Iris contiene **150 flores de 3 especies**, con 50 flores por especie. |
| **Instancia, muestra o ejemplo** | Una observación individual. En Iris, cada fila representa una flor. |
| **Atributo, característica o variable de entrada** | Una medida que se usa para describir una flor. Iris tiene cuatro atributos: largo y ancho del sépalo, y largo y ancho del pétalo. |
| **Clase** | Categoría que el clasificador intenta identificar. Las clases de Iris son *setosa*, *versicolor* y *virginica*. |
| **Etiqueta real o variable objetivo (`target`)** | La clase conocida de una instancia. Iris representa las especies con los códigos `0`, `1` y `2`, respectivamente. |
| **`X` e `y`** | Por convención, `X` contiene las medidas de las flores y `y` contiene sus etiquetas. Cada fila de `X` corresponde a la etiqueta de la misma posición en `y`. |
| **Arreglo de NumPy (`ndarray`)** | Estructura que guarda los datos numéricos. Las medidas forman un arreglo de dos dimensiones y las etiquetas, uno de una dimensión. |
| **`iris.data`, `iris.target`, `iris.target_names`** | Contienen, respectivamente, las medidas, los códigos de especie y los nombres de las especies. |
| **`.shape`** | Indica el tamaño de un arreglo. En una tabla de medidas, `.shape[0]` da el número de filas y `.shape[1]` el número de columnas. En Iris, `iris.data.shape` es `(150, 4)`. |
| **Frecuencia u ocurrencia** | Número de veces que aparece una clase. `np.unique(y, return_counts=True)` devuelve los códigos distintos y sus conteos en el mismo orden. |

## Partición de los datos

| Término | Significado en el ejercicio |
|---|---|
| **Hold-out** | Método que reserva una parte de los datos para evaluar el modelo con ejemplos que quedaron fuera de su ajuste. |
| **Partición externa** | Primera división. Con una proporción 80/20, `X_training` y `y_training` contienen **120 flores** disponibles para entrenar y validar. `X_test` y `y_test` contienen **30 flores** para la prueba final. |
| **Partición interna** | Segunda división. Toma las 120 flores de la partición externa y las reparte en **96 de entrenamiento** y **24 de validación**. |
| **Conjunto de entrenamiento (`X_train`, `y_train`)** | Medidas y etiquetas que se usan para ajustar el clasificador. |
| **Conjunto de validación (`X_val`, `y_val`)** | Medidas y etiquetas que se usan para comparar modelos o configuraciones durante el desarrollo. |
| **Conjunto de prueba (`X_test`, `y_test`)** | Medidas y etiquetas reservadas para evaluar el modelo elegido. También se llama conjunto de *test*. |
| **`train_test_split()`** | Función de scikit-learn que divide los arreglos y mantiene cada flor emparejada con su etiqueta. |
| **`test_size`** | Proporción que se reserva en una llamada a `train_test_split()`. Con `0.2` se separa un 20 %. En la partición interna, ese 20 % corresponde a validación. |
| **`random_state`** | Semilla que permite repetir la misma división con los mismos datos y parámetros. En el ejercicio se usaron valores como `0` y `42`. |

## Preprocesamiento

| Término | Significado en el ejercicio |
|---|---|
| **Preprocesamiento** | Preparación de las medidas antes de entregarlas al clasificador. |
| **Estandarización** | Transformación que resta la media de cada columna y divide por su desviación estándar. Se aplica a las medidas de `X`. |
| **Media** | Promedio de los valores de una columna. |
| **Desviación estándar** | Medida de cuánto se dispersan los valores alrededor de la media. |
| **`StandardScaler`** | Clase de scikit-learn que calcula y aplica la estandarización. |
| **`fit_transform(X_train)`** | Calcula las estadísticas de entrenamiento y las usa para transformar `X_train` en el mismo paso. |
| **`transform(X_val)` y `transform(X_test)`** | Aplican a validación y prueba las estadísticas calculadas con `X_train`. |

## Clasificación, predicción y evaluación

| Término | Significado en el ejercicio |
|---|---|
| **Aprendizaje supervisado** | Entrenamiento con ejemplos que incluyen su respuesta correcta. En Iris, las respuestas son las especies reales. |
| **Clasificación** | Tarea de predecir una categoría. Iris es un problema de **clasificación multiclase** porque hay tres especies posibles. |
| **Clasificador o modelo** | Objeto que, una vez ajustado, puede asignar una clase a cada flor a partir de los datos que recibe. |
| **Clasificador de referencia (`DummyClassifier`)** | Modelo que usa una regla simple para obtener un resultado de referencia con el cual comparar otros clasificadores. |
| **Estrategia `prior`** | Opción de `DummyClassifier` que predice siempre la clase más frecuente de `y_train`. Esa clase se llama **clase mayoritaria**. |
| **`SVC`** | Clasificador de máquinas de vectores de soporte de scikit-learn. Usa las medidas de las flores para aprender a distinguir las especies. |
| **Hiperparámetro `C`** | Valor que se elige antes de entrenar `SVC`. Controla la penalización de los errores de entrenamiento. En el ejercicio se usó `C=0.5`. |
| **Ajuste o entrenamiento (`fit`)** | Paso en el que el clasificador aprende a partir de los datos. Por ejemplo, `svc.fit(X_train, y_train)`. |
| **Predicción (`predict`)** | Clase que el modelo asigna a cada instancia. `y_pred_val` guarda las predicciones de validación y `y_pred_test`, las de prueba. |
| **Exactitud (`accuracy`)** | Proporción de predicciones correctas. Por ejemplo, 24 aciertos en 30 flores dan `24 / 30 = 0.8`, equivalente al **80 %**. |
| **`score(X, y)`** | Método que, en los clasificadores usados en el ejercicio, devuelve la exactitud sobre las medidas `X` y las etiquetas reales `y`. |
| **`np.round(valor, 4)`** | Redondea un valor a cuatro decimales. Multiplicar la exactitud por `100` antes de redondearla la convierte en porcentaje. |
